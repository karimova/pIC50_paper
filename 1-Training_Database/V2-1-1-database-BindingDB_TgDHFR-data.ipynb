## 6.1. Step 1: Remove Outliers

```python
import pandas as pd

# Function to remove outliers using the IQR method
def remove_outliers_iqr(df, column):
    Q1 = df[column].quantile(0.25)
    Q3 = df[column].quantile(0.75)
    IQR = Q3 - Q1
    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR
    return df[(df[column] >= lower_bound) & (df[column] <= upper_bound)]

# Apply the outlier removal function to each group
grouped_dupl_list_no_outliers = grouped_dupl_list.groupby('Smiles').apply(lambda group: remove_outliers_iqr(group, 'IC50 (nM)')).reset_index(drop=True)

# Display the cleaned DataFrame without outliers
grouped_dupl_list_no_outliers.head(10)
```

## 6.2. Step 2: Calculate Average IC50:

```python
# Calculate the average of log_IC50 for each group in grouped_dupl_tg_list_no_outliers
average_ic50_df = grouped_dupl_list_no_outliers.groupby(['Smiles'], as_index=False)['IC50 (nM)'].mean()

# Rename the column for clarity
average_ic50_df.rename(columns={'IC50 (nM)': 'IC50 (nM) Avg'}, inplace=True)

# Display the resulting DataFrame with BindingDB Target Chain Sequence, Ligand SMILES, and their average log_IC50 values
average_ic50_df
```
